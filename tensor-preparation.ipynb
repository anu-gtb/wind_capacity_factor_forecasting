```python
import pandas as pd
import numpy as np
import torch
```

```python
def create_sliding_windows(data,lookback=12,horizon=1):
    x,y=[],[]
    for i in range(len(data)-lookback-horizon+1):
        x.append(data[i:(i+lookback),:])
        y.append(data[(i+lookback):(i+lookback+horizon),:])
    return np.array(x),np.array(y)
```

```python
df_matrix=pd.read_pivot=pd.read_csv('/kaggle/input/datasets/anubhashukla/tensor-data-creation/grid_timeseries.csv',parse_dates=['Timestamp']).pivot(index='Timestamp',columns='Node',values='Power')
data_matrix=df_matrix.values
```

```python
x,y=create_sliding_windows(data_matrix,lookback=12,horizon=1)
```

```python
split=int(len(x)*0.8)
x_train,x_test=torch.tensor(x,dtype=torch.float32),torch.tensor(x[split:],dtype=torch.float32)
y_train,y_test=torch.tensor(y,dtype=torch.float32),torch.tensor(y[split:],dtype=torch.float32)
```

```python
x_train=x_train.permute(0,2,1).unsqueeze(1)
x_test=x_test.permute(0,2,1).unsqueeze(1)
y_train=y_train.squeeze(1)
y_test=y_test.squeeze(1)
```

```python
torch.save({'x_train':x_train,'y_train':y_train,'x_test':x_test,'y_test':y_test},'gnn_tensors.pt')
print(f'Tensors shape:{x_train.shape}')
```
