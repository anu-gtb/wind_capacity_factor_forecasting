```python
chunk_size=100000
chunks=[]

start_date='2022-01-01 00:00:00'
end_date='2022-01-31 23:00:00'

df=pd.read_csv('C:\\Users\\DELL\\Documents\\STATISTICS\\wind_capacity_factor_2022.csv',chunksize=chunk_size)
for chunk in df:
  chunk['time']=pd.to_datetime(chunk['time'])
  
  time_mask=(chunk['time']>=start_date)&(chunk['time']<=end_date)
  sliced_chunk=chunk[time_mask]
  
  if not sliced_chunk.empty:
    chunks.append(sliced_chunk)

df_sliced=pd.concat(chunks,ignore_index=True)
df_sliced.to_csv('sliced_data.csv',index=False)
```
