```python
import numpy as np
from statsmodels.stats.diagnostic import acorr_ljungbox
```

```python
data=np.load('/kaggle/input/datasets/anubhashukla/evaluated-data/model_evaluations.npz')
actuals,gnn,var=data['actuals'],data['gnn'],data['var']
```

```python
hourly_deltas=np.abs(np.diff(actuals,axis=0).sum(axis=1))
extreme_event_idx=int(np.argmax(hourly_deltas))
```

```python
print('Accuracy Stress Test(EXTREME EVENT HOUR)')
print(f"Spatio-Temporal Graph Network MSE: {np.mean((actuals[extreme_event_idx] - gnn[extreme_event_idx])**2)}")
print(f"Vector Autoregression Baseline MSE: {np.mean((actuals[extreme_event_idx] - var[extreme_event_idx])**2)}")
```

```python
gnn_residuals=(actuals-gnn).flatten()
diagnostic_df=acorr_ljungbox(gnn_residuals, lags=[10], return_df=True)
p_val = diagnostic_df['lb_pvalue'].values[0]
print(f"\nLjung-Box Autocorrelation Verification Test P-Value: {p_val}")
```
