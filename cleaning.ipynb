```python
df['time']=pd.to_datetime(df['time'])
df['Node']='Node_('+df['latitude'].astype(str)+','+df['longitude'].astype(str)+')'
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.sort_values(by=['Node','time'])
```

```python
df_matrix=df.pivot_table(index='time',columns='Node',values='Vestas_v110_2000MW',aggfunc='mean')
```

```python
node_coordinates_map=df[['Node','latitude','longitude']].drop_duplicates()
```

```python
df_matrix.to_csv('temp_matrix.csv')
node_coordinates_map.to_csv('node_coordinates.csv',index=False)
```
